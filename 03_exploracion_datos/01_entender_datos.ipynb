## Valores nulos 
Luego de ver la estructura y de idenfiticar las tablas que tienen valores nulos, se va identificar en que columnas están los valores nulos

```python
for nombre, df in tabla_archivos.items():
    cantidad_registros = df.shape[0] 
    valores_nulos = df.isnull().sum()
    nulos = valores_nulos[valores_nulos > 0]

    if len(nulos) > 0:
        #cantidad_nulos = nulos.sum()
        print(f'\n{nombre.upper()}:')
        #print(f'TOTAL {cantidad_nulos} / {cantidad_registros}')
        for columna, cantidad in nulos.items():
            print(f'La columna "{columna}": Tiene {cantidad} valores vacios de {cantidad_registros} registros')
```

## Duplicados
Voy a realizar la identificación de los valores duplicados

```python
for nombre, df in tabla_archivos.items():
    duplicados = df.duplicated().sum()

    if duplicados > 0:
        print(f'\n{nombre.upper()}: {duplicados} duplicados')
    else:
        print(f'\n{nombre.upper()}: No tiene registros duplicados')
```
